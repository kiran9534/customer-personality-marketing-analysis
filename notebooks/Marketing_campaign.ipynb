The business problem

We'll pretend we're analysts working for a consumer/e-commerce company.

The company has customer data but wants to understand:

Who are our customers, what do they buy, how do they purchase, and which types of customers respond to marketing campaigns?

Then we'll answer several business questions.

Customer questions
Who are the high-value customers?
How does income relate to spending?
Which products generate the most spending?
Do customers with children spend differently?
How does education relate to spending?
Marketing questions
Which campaigns have the highest acceptance?
What characteristics are associated with campaign response?
Do high-spending customers respond more frequently?
Which purchasing channel is most commonly used?
Are website visits associated with purchases?
Segmentation question

Can we divide customers into meaningful groups based on their characteristics and purchasing behavior?

```python
df_clean["Age"] = 2014 - df_clean["Year_Birth"]
```

```python
df_clean["Age"].describe()
```

```python
df_clean["Total_Spending"] = (
    df_clean["MntWines"]
    + df_clean["MntFruits"]
    + df_clean["MntMeatProducts"]
    + df_clean["MntFishProducts"]
    + df_clean["MntSweetProducts"]
    + df_clean["MntGoldProds"]
)
```

```python
df_clean["Total_Spending"].describe()
```

```python
df_clean["Total_Purchases"] = (
    df_clean["NumWebPurchases"]
    + df_clean["NumCatalogPurchases"]
    + df_clean["NumStorePurchases"]
)
```

```python
df_clean["Total_Purchases"].describe()
```

```python
df_clean["Total_Campaign_Accepted"] = (
    df_clean["AcceptedCmp1"]
    + df_clean["AcceptedCmp2"]
    + df_clean["AcceptedCmp3"]
    + df_clean["AcceptedCmp4"]
    + df_clean["AcceptedCmp5"]
)
```

```python
df_clean["Children"] = (df_clean["Kidhome"] + df_clean["Teenhome"])
```

```python
#df_clean["Parental_status"] = df_clean["Children"].apply( lamda x : "Parent" if x > 0 else "Non-Parent")

df_clean["Parental_Status"] = df_clean["Children"].apply(
    lambda x: "Parent" if x > 0 else "Non-Parent"
)
```

```python
df_clean["Parental_Status"].value_counts()
```

```python
df_clean.columns
```

```python
df_c = df_clean.copy()
```

```python
df_c["Dt_Customer"] = pd.to_datetime(
    df_c["Dt_Customer"],
    dayfirst=True,
    errors="coerce"
)
```

```python
df_c["Dt_Customer"].dtype
```

```python
reference_date = pd.Timestamp("2014-12-31")

df_c["Customer_Tenure_Days"] = (
    reference_date - df_c["Dt_Customer"]
).dt.days

df_c["Customer_Tenure_Years"] = (
    df_c["Customer_Tenure_Days"] / 365.25
)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize = (10, 5))
plt.hist(
    df_c["Age"].dropna(),
    bins= 20,
    edgecolor= "black"

)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Nmber of Customers")
plt.show()
```

```python
 plt.figure(figsize=(10, 5))

plt.hist(
    df_c["Income_Analysis"].dropna(),
    bins=30,
    edgecolor="black"
)

plt.title("Customer Income Distribution")
plt.xlabel("Income")
plt.ylabel("Number of Customers")

plt.show()
```

```python
spending_by_category = {
    "Wines": df_c["MntWines"].sum(),
    "Fruits": df_c["MntFruits"].sum(),
    "Meat": df_c["MntMeatProducts"].sum(),
    "Fish": df_c["MntFishProducts"].sum(),
    "Sweets": df_c["MntSweetProducts"].sum(),
    "Gold": df_c["MntGoldProds"].sum()
}

spending_by_category
```

```python
spending_df = pd.DataFrame(
    list(spending_by_category.items()),
    columns=["Category", "Total_Spending"]
)

spending_df
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    spending_df["Category"],
    spending_df["Total_Spending"]
)

plt.title("Total Customer Spending by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Spending")

plt.xticks(rotation=0)
plt.show()
```

```python
channel_purchases = {
    "Web": df_c["NumWebPurchases"].sum(),
    "Catalog": df_c["NumCatalogPurchases"].sum(),
    "Store": df_c["NumStorePurchases"].sum(),
    "Deals": df_c["NumDealsPurchases"].sum()
}

channel_df = pd.DataFrame(
    list(channel_purchases.items()),
    columns=["Channel", "Total_Purchases"]
)

channel_df
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    channel_df["Channel"],
    channel_df["Total_Purchases"]
)

plt.title("Customer Purchases by Channel")
plt.xlabel("Purchase Channel")
plt.ylabel("Total Purchases")

plt.show()
```

```python
campaign_response = {
    "Campaign 1": df_c["AcceptedCmp1"].sum(),
    "Campaign 2": df_c["AcceptedCmp2"].sum(),
    "Campaign 3": df_c["AcceptedCmp3"].sum(),
    "Campaign 4": df_c["AcceptedCmp4"].sum(),
    "Campaign 5": df_c["AcceptedCmp5"].sum()
}

campaign_df = pd.DataFrame(
    list(campaign_response.items()),
    columns=["Campaign", "Accepted_Customers"]
)

campaign_df
```

```python
campaign_df["Response_Rate_%"] = (
    campaign_df["Accepted_Customers"] / len(df_c) * 100
).round(2)

campaign_df
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    campaign_df["Campaign"],
    campaign_df["Response_Rate_%"]
)

plt.title("Marketing Campaign Response Rate")
plt.xlabel("Campaign")
plt.ylabel("Response Rate (%)")

plt.show()
```
